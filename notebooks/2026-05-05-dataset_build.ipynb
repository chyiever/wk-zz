# Dataset Build

This notebook is split into four parts: environment setup, parameter configuration, build execution, and result inspection.

Edit only the parameter cell for normal use.

```python
from pathlib import Path
import sys

repo_root = Path.cwd().parent if (Path.cwd() / 'src').exists() is False else Path.cwd()
src_root = repo_root / 'src'
if str(src_root) not in sys.path:
    sys.path.insert(0, str(src_root))

from dataset_build import BuildConfig, build_all_datasets, build_single_noise_domain
print('dataset_build imported')
```

## Parameters

Tune this cell only. The rest of the notebook is execution and inspection.

```python
# Build mode: 'all' builds F130, F130A, and F130C together; 'single' builds one selected noise domain.
build_mode = 'all'
single_noise_dataset = 'F130'

# Input and output roots should point to the repository root.
data_root = repo_root / 'data'
output_root = repo_root / 'data' / 'dataset_build_260518'

# Split and randomization controls.
train_ratio = 0.7
random_seed = 260505
show_progress = True

# Shared signal settings.
target_sample_rate_hz = 500_000.0
window_duration_s = 0.02
t_min_s = -0.0080
t_max_s = 0.0050
t_step_s = 0.0001
w_min = 0.5
w_max = 1.5

# Flow-noise group sizes for time-neighborhood splitting.
flow_group_sizes = {
    'F130': 10,
    'F130A': 20,
    'F130C': 10,
}

# Which domains to build in 'all' mode.
selected_noise_datasets = ('F130', 'F130A', 'F130C')

config = BuildConfig(
    data_root=data_root,
    output_root=output_root,
    target_sample_rate_hz=target_sample_rate_hz,
    train_ratio=train_ratio,
    random_seed=random_seed,
    window_duration_s=window_duration_s,
    t_min_s=t_min_s,
    t_max_s=t_max_s,
    t_step_s=t_step_s,
    w_min=w_min,
    w_max=w_max,
    flow_group_sizes=flow_group_sizes,
    noise_datasets=selected_noise_datasets,
)

config
```

```python
print('Build preview')
print(f'build_mode = {build_mode}')
print(f'data_root = {config.data_root}')
print(f'output_root = {config.output_root}')
print(f'train_ratio = {config.train_ratio}')
print(f'random_seed = {config.random_seed}')
print(f'target_sample_rate_hz = {config.target_sample_rate_hz}')
print(f'window_duration_s = {config.window_duration_s}')
print(f't_range = [{config.t_min_s}, {config.t_max_s}] step {config.t_step_s}')
print(f'w_range = [{config.w_min}, {config.w_max}]')
print(f'flow_group_sizes = {config.flow_group_sizes}')
print(f'noise_datasets = {config.noise_datasets}')
```

```python
if build_mode == 'all':
    result = build_all_datasets(config, show_progress=show_progress)
elif build_mode == 'single':
    result = build_single_noise_domain(
        config,
        noise_dataset=single_noise_dataset,
        show_progress=show_progress,
    )
else:
    raise ValueError(f"Unsupported build_mode: {build_mode!r}. Use 'all' or 'single'.")

result
```

## Example: downsample one BK14 500K sample

This cell converts one fixed BK14 500K source file to 200 kHz and saves the result to a local preview folder for inspection.

The downsampling path removes the linear trend before `resample_poly` and uses `padtype='line'` to suppress the 100 kHz Nyquist-edge artifact and edge transients.

```python
from pathlib import Path

import numpy as np

from bk_analysis.io import load_signal_record
from dataset_build.resample import resample_to_rate
from dataset_build.writers import write_npz

example_source = repo_root / 'data' / 'BK14-0.06S(Generation method)' / 'corrosion' / 'BK14-FIP-500K-20260325T101142.414.npz'
example_output_root = repo_root / 'data' / 'downsample_preview'
example_output_root.mkdir(parents=True, exist_ok=True)
example_output_path = example_output_root / f'{example_source.stem}-sr200k.npz'

with np.load(example_source, allow_pickle=True) as raw:
    original_timestamp = float(raw['timestamp'])
    original_starttime = str(raw['starttime'].tolist())
    original_arrival_time = str(raw['arrival_time'].tolist())
    original_sample_type = str(raw['type'].tolist())
    original_data_info = raw['data_info'].item() if 'data_info' in raw else {}

record = load_signal_record(example_source, method='corrosion')
resampled_signal, resample_action = resample_to_rate(record.signal, record.sample_rate, 200_000.0)

data_info = dict(original_data_info) if isinstance(original_data_info, dict) else {}
data_info.update(
    {
        'example_source_file': example_source.name,
        'example_target_sample_rate_hz': 200_000.0,
        'example_resample_action': resample_action,
        'example_note': 'single BK14 500K sample downsampled to 200K for manual inspection',
    }
)

write_npz(
    output_path=example_output_path,
    signal_values=resampled_signal,
    sample_rate_hz=200_000.0,
    starttime=original_starttime,
    arrival_time=original_arrival_time,
    timestamp=original_timestamp,
    sample_type=original_sample_type,
    data_info=data_info,
)

f = np.fft.rfftfreq(len(resampled_signal), d=1 / 200_000.0)
m = np.abs(np.fft.rfft(resampled_signal))
top = np.argsort(m)[-8:][::-1]

print(f'source: {example_source}')
print(f'output: {example_output_path}')
print(f'original_len = {len(record.signal)}')
print(f'resampled_len = {len(resampled_signal)}')
print(f'resample_action = {resample_action}')
print('top spectral peaks (Hz, magnitude):')
for idx in top:
    print(f'  {float(f[idx]):.1f}, {float(m[idx]):.6f}')
```

## Example: upsample one BK14 200K sample

This cell upsamples one fixed BK14 200K source file to 500 kHz and saves the result to a local preview folder for comparison.

```python
from pathlib import Path

import numpy as np

from bk_analysis.io import load_signal_record
from dataset_build.resample import resample_to_rate
from dataset_build.writers import write_npz

upsample_source = repo_root / 'data' / 'BK14-0.06S(Generation method)' / 'corrosion' / 'BK14-FIP-200K-20260323T120121.475.npz'
upsample_output_root = repo_root / 'data' / 'upsample_preview'
upsample_output_root.mkdir(parents=True, exist_ok=True)
upsample_output_path = upsample_output_root / f'{upsample_source.stem}-sr500k.npz'

with np.load(upsample_source, allow_pickle=True) as raw:
    original_timestamp = float(raw['timestamp'])
    original_starttime = str(raw['starttime'].tolist())
    original_arrival_time = str(raw['arrival_time'].tolist())
    original_sample_type = str(raw['type'].tolist())
    original_data_info = raw['data_info'].item() if 'data_info' in raw else {}

record = load_signal_record(upsample_source, method='corrosion')
upsampled_signal, resample_action = resample_to_rate(record.signal, record.sample_rate, 500_000.0)

data_info = dict(original_data_info) if isinstance(original_data_info, dict) else {}
data_info.update(
    {
        'example_source_file': upsample_source.name,
        'example_target_sample_rate_hz': 500_000.0,
        'example_resample_action': resample_action,
        'example_note': 'single BK14 200K sample upsampled to 500K for manual comparison',
    }
)

write_npz(
    output_path=upsample_output_path,
    signal_values=upsampled_signal,
    sample_rate_hz=500_000.0,
    starttime=original_starttime,
    arrival_time=original_arrival_time,
    timestamp=original_timestamp,
    sample_type=original_sample_type,
    data_info=data_info,
)

f = np.fft.rfftfreq(len(upsampled_signal), d=1 / 500_000.0)
m = np.abs(np.fft.rfft(upsampled_signal))
top = np.argsort(m)[-8:][::-1]

print(f'source: {upsample_source}')
print(f'output: {upsample_output_path}')
print(f'original_len = {len(record.signal)}')
print(f'upsampled_len = {len(upsampled_signal)}')
print(f'resample_action = {resample_action}')
print('top spectral peaks (Hz, magnitude):')
for idx in top:
    print(f'  {float(f[idx]):.1f}, {float(m[idx]):.6f}')

print('first10:', np.round(upsampled_signal[:10], 3).tolist())
print('last10:', np.round(upsampled_signal[-10:], 3).tolist())
print('max diff first50:', float(np.max(np.abs(np.diff(upsampled_signal)[:50]))))
print('max diff last50:', float(np.max(np.abs(np.diff(upsampled_signal)[-50:]))))
```
