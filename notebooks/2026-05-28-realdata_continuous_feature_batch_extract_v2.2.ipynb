## v2.2 架构说明```主进程(GPU):  加载文件N → 批量GPU STFT(200窗口/批) → 写入共享内存 → 加载文件N+1 → ...       ↕ SharedMemory (signal_pre + STFT结果)  预加载线程: 后台加载下一文件，与当前文件的特征计算并行持久Worker1-14(CPU):  从共享内存读取信号切片 + STFT切片 → butter_filter + ridge + ISTFT + wavelet + 特征计算 → 返回结果```**关键保证：所有优化不改变特征计算公式，仅改变计算调度方式。**
